# Web scraper

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url = 'https://webscraper.io/test-sites/pagination'

page = requests.get(url)
```

```python
soup = BeautifulSoup(page.text, "lxml")

info = soup.find_all('p', class_='description')
```

```python
print(info[0].text)
print('='*40)
# .text works for single output for multiple loop through the list
for i in info:
    print(i.text)
```
